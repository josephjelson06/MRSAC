```python
import os

# Paths
yolo_labels_path = r"C:\project_mrsac\DATASET\train\labels"  # Original annotations
seg_labels_path = r"C:\project_mrsac\DATASET\train\labels_seg"  # Converted annotations

os.makedirs(seg_labels_path, exist_ok=True)

def merge_classes(label_path, save_path, target_class=0):
    for file in os.listdir(label_path):
        with open(os.path.join(label_path, file), "r") as f:
            lines = f.readlines()

        new_annotations = []
        for line in lines:
            values = line.strip().split()
            values[0] = str(target_class)  # Change class to 0
            new_annotations.append(" ".join(values))

        with open(os.path.join(save_path, file), "w") as f:
            f.write("\n".join(new_annotations))

merge_classes(yolo_labels_path, seg_labels_path)
print("✅ All farm boundaries are now labeled as a single class!")
```

```python
from ultralytics import YOLO
import torch

# ✅ Check if CUDA (GPU) is available
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# ✅ Load YOLOv8 segmentation model
model = YOLO("yolov8s-seg.pt")  # Using YOLOv8 Small (Segmentation)
```

```python
# ✅ Train the segmentation model
model.train(
    data=r"C:\project_mrsac\DATASET\data.yaml",  # Path to your dataset
    epochs=10,  # Number of training epochs
    imgsz=640,  # Image size
    batch=8,  # Adjust based on your system
    device=device,  # Use GPU if available
    workers=4,  # Number of CPU threads for loading data
    project="runs/farm_segmentation",
    name="exp1"
)
```
