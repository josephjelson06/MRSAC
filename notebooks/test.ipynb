```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
from ultralytics import YOLO

def detect_objects_batch(image_folder, model_path=r"C:\project_mrsac\runs\detect\train8\weights\best.pt"):
    # ✅ Load trained YOLO model
    model = YOLO(model_path)
    
    # ✅ Get all image file paths (filter for common image formats)
    image_extensions = ('*.jpg', '*.jpeg', '*.png', '*.bmp', '*.tiff')
    image_paths = []
    for ext in image_extensions:
        image_paths.extend(glob.glob(os.path.join(image_folder, ext)))
    
    if not image_paths:
        print("Error: No images found in the folder.")
        return
    
    detections = []
    
    for image_path in image_paths:
        # ✅ Read image
        image = cv2.imread(image_path)
        
        if image is None:
            print(f"Error: Image not loaded - {image_path}")
            continue
        
        # ✅ Run detection
        results = model.predict(image, save=True, save_txt=True)  # Explicitly enable saving
        
        # ✅ Process results
        for r in results:
            boxes = r.boxes.xyxy.cpu().numpy()
            classes = r.boxes.cls.cpu().numpy()
            confidences = r.boxes.conf.cpu().numpy()
            
            for box, cls, conf in zip(boxes, classes, confidences):
                x1, y1, x2, y2 = map(int, box)
                label = f"{model.names[int(cls)]} ({conf:.2f})"
                cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
                cv2.putText(image, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
                detections.append((label, conf))
        
        # ✅ Convert BGR to RGB
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        
        # ✅ Display image
        plt.figure(figsize=(8, 6))
        plt.imshow(image_rgb)
        plt.axis("off")
        plt.title(os.path.basename(image_path))
        plt.show(block=True)
    
    # ✅ Plot detection confidence scores
    if detections:
        labels, confs = zip(*detections)
        plt.figure(figsize=(10, 5))
        plt.barh(labels, confs, color='blue')
        plt.xlabel("Confidence Score")
        plt.ylabel("Detected Classes")
        plt.title("Detection Confidence Scores")
        plt.show()
    
    print("Batch processing complete.")
    
# Example usage
detect_objects_batch(r"C:\project_mrsac\DATASET\test\images")

```
